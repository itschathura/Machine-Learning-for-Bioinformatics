```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#How many patients belongs to each cacer type - Bar Chart

sns.countplot(data=df, x = "TumorType")
plt.title("No of Patients by Tumor Type")
plt.xlabel("Tumor type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Box Plot - survival by tumor type

sns.boxplot(
    data = df,
    x = "TumorType",
    y = "SurvivalInMonths"
)


plt.title("No of Patients by Tumor Type")
plt.show()
```

```python
#Scatter Plot

sns.scatterplot(
    data=df,
    x="MutationCount",
    y="SurvivalInMonths",
    hue="TreatmentResponse"
)

plt.title("Mutation Count VS Survival")
plt.xlabel("Mutation Count")
plt.ylabel("Survival (Months)")
plt.show()
```

```python
#Heatmap

numeric_df = df[
    ["Age", "SurvivalInMonths", "MutationCount"]
]

correlation = numeric_df.corr()
sns.heatmap(
    correlation,
    annot = True,
    cmap = "coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#Heatmap

numeric_df = df[
    ["Age", "SurvivalInMonths", "MutationCount"]
]

correlation = numeric_df.corr()
sns.heatmap(
    correlation,
    annot = True,
    cmap = "YlGn" #'viridis', 'plasma', 'inferno', 'coolwarm', 'YlGn'
)

plt.title("Correlation Heatmap")
plt.show()
```
