```python
#create a DNA sequence

dna = "CTTTCTGTCCCGCCCTTCCTCTGACTGTGTCTTGATTTCCTATTCTGAGAGGCTATTGCTCAGCGGTTTCCCACCGCGCGCCCGCCCAGCGCCTGCATC"
print(dna)
```

```python
type(dna)
```

```python
print(len(dna))
```

```python
#DNA -> RNA

rna = dna.replace("T", "U")
print("DNA : ",dna)
print("RNA : ",rna)
```

```python
#Count

print("A : ", dna.count("A"))
print("T : ", dna.count("T"))
print("G : ", dna.count("G"))
print("C : ", dna.count("C"))
```

```python
import matplotlib.pyplot as plt
A = dna.count("A")
T = dna.count("T")
G = dna.count("G")
C = dna.count("C")

nucleotides = ["A", "T", "G", "C"]
counts = [A, T, G, C]

plt.bar(nucleotides, counts)
plt.title("Nucleotide Count")
plt.xlabel("Nucleotides")
plt.ylabel("Count")
plt.show()
```

```python
#GC content

gc_count = dna.count("G") + dna.count("C")
gc_content = (gc_count / len(dna)) * 100
print("GC Count : ", gc_content, "%")
```

```python
#DNA -> RNA

rna = dna.replace("T", "U")
print("DNA : ",dna)
print("RNA : ",rna)
```

```python
#TRANSLATE

complement = dna.translate(str.maketrans("ATGC", "TACG"))
print(complement)
```

```python
#reverse complement

reverse_complement = complement[::-1]
#take the seq ifrom begining to end with astep of -1
print(reverse_complement)
```
